## ***Q1. Student Assessment Data Cleaning***

#### **Question**

A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing.

**Dataset:** `student_assessment_dirty.csv`

## ***1. Load and Inspect the Dataset***


````python
import pandas as pd

df = pd.read_csv("student_assessment_dirty.csv", keep_default_na=False)

print(df.head())
print(df.info())
print(df.isnull().sum())

#Code Explanation: Here, I loaded the CSV file and used head(), info(), and isnull().sum() to understand the dataset and check for missing values.
````

## ***2. Standardize Department Names***

````python
df["Department"] = df["Department"].str.strip().str.lower()

df["Department"] = df["Department"].replace({
    "it": "IT",
    "i.t.": "IT",
    "hr": "HR",
    "finance": "Finance"
})

print(df["Department"])


#Code Explanation: Here, I standardized department names so that different forms such as IT, it, and I.T. are represented consistently.
````

## ***3. Remove Exact Duplicate Recordsv***

````python
before = len(df)

df = df.drop_duplicates()

removed = before - len(df)

print("Records removed:", removed)

#Code Explanation: Here, I used drop_duplicates() to remove the exact duplicate record of student S102.
````

## ***4. Convert Marks into Numeric Format***

The dataset contains different types of values in the Marks column:

Blank → mark not entered

Absent → student was absent

NA → mark not entered

88 marks → should become 88

Numeric values → converted to numbers

````python
import numpy as np

conditions = [
    df["Marks"].eq("Absent"),
    df["Marks"].isin(["", "NA"])
]

choices = ["Absent", "Not Entered"]

df["Mark_Status"] = np.select(
    conditions,
    choices,
    default="Entered"
)

df["Marks"] = df["Marks"].str.extract(
    r"(\d+(?:\.\d+)?)",
    expand=False
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df)

#Code Explanation: Here, I separated the status of missing marks and converted values such as 88 marks into numeric values.
````

### ***Important Result***

#### **For example:**
```text
88 marks → 88
Absent   → NaN
NA       → NaN
blank    → NaN
```
The Mark_Status column helps us distinguish Absent from marks that were not entered.

## ***5. Investigate Missing Marks***

````python
print(df[df["Marks"].isna()][
    ["Student_ID", "Student_Name", "Marks", "Mark_Status"]
])
````

### ***Explanation***

The missing marks should not all be treated in the same way.

Meera is **Absent**.

Therefore, these values should be investigated before deleting them.

### ***Risk of deleting missing values***

Deleting missing values without investigating them may remove useful student records and can lead to incorrect analysis.

## ***6. Identify Marks Outside the Valid Range***

The valid mark range is 0 to 100.

````python
invalid_marks = df[
    df["Marks"].notna() &
    ~df["Marks"].between(0, 100)
]

print(invalid_marks)
````

### ***Explanation***

Kabir has a mark of 105, which is outside the valid range of 0–100. This value should be investigated before changing or removing it because it could be a data-entry error.

````python
print("Records removed:", removed)
print("Missing marks remaining:", df["Marks"].isna().sum())
print("Invalid marks identified:", len(invalid_marks))
````
